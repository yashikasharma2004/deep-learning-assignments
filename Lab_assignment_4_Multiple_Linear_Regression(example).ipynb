```python
import pandas as pd
import numpy as np

try:
    data = pd.read_csv("multiple_linear_regression_dataset.csv")
except:
    print("Error: File not found.")

X = data[["age", "experience"]].values
y = data["income"].values

n_features = X.shape[1]
w = np.zeros(n_features)
b = 0.0

def predict(X, w, b):
    return X.dot(w) + b

def mean_squared_error(y, y_hat):
    return ((y_hat - y) ** 2).mean()

def compute_gradients(X, y, y_hat):
    N = len(y)
    dw = (2/N) * X.T.dot(y_hat - y)
    db = (2/N) * (y_hat - y).sum()
    return dw, db

def update_parameters(w, b, dw, db, lr):
    w = w - lr * dw
    b = b - lr * db
    return w, b

lr = 0.00001
epochs = 2000

for epoch in range(epochs):
    y_hat = predict(X, w, b)
    loss = mean_squared_error(y, y_hat)
    dw, db = compute_gradients(X, y, y_hat)
    w, b = update_parameters(w, b, dw, db, lr)

    if epoch % 500 == 0:
        print(f"Epoch {epoch}, Loss: {loss}")

print("\n--- Final Results ---")
print("Final Weights:", w)
print("Final Bias:", b)

new_candidate = np.array([35, 8])
predicted_income = new_candidate.dot(w) + b
print(f"Predicted Income: {predicted_income:.2f}")
```

```python
import pandas as pd
import numpy as np

column_names = ["Sex", "Length", "Diameter", "Height", "Whole_weight",
                "Shucked_weight", "Viscera_weight", "Shell_weight", "Rings"]
data = pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/abalone/abalone.data", names=column_names)

data['y'] = data['Rings'] + 1.5

features = ["Length", "Diameter", "Whole_weight"]
X = data[features].values
y = data['y'].values.reshape(-1, 1)

split_idx = int(0.8 * len(X))
X_train, X_test = X[:split_idx], X[split_idx:]
y_train, y_test = y[:split_idx], y[split_idx:]

train_mean = np.mean(X_train, axis=0)
train_std = np.std(X_train, axis=0)
X_train = (X_train - train_mean) / train_std
X_test = (X_test - train_mean) / train_std

w = np.random.randn(3, 1) * 0.01
b = 0.0

def forward(X, w, b):
    return np.dot(X, w) + b

def mse(y, y_hat):
    return np.mean((y - y_hat)**2)

def grad_w(X, y, y_hat):
    return (2/len(y)) * np.dot(X.T, (y_hat - y))

def grad_b(y, y_hat):
    return (2/len(y)) * np.sum(y_hat - y)

lr = 0.01
epochs = 1000

for epoch in range(epochs):
    y_hat = forward(X_train, w, b)
    loss = mse(y_train, y_hat)

    dw = grad_w(X_train, y_train, y_hat)
    db = grad_b(y_train, y_hat)

    w -= lr * dw
    b -= lr * db

    if epoch % 200 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")

y_hat_test = forward(X_test, w, b)
test_mse = mse(y_test, y_hat_test)
test_mae = np.mean(np.abs(y_test - y_hat_test))

print("\n--- Final Evaluation ---")
print(f"Test MSE: {test_mse:.4f}")
print(f"Test MAE: {test_mae:.4f}")

print("\nSample Predictions:")
for i in range(5):
    print(f"True: {y_test[i][0]}, Predicted: {y_hat_test[i][0]:.2f}, Abs Error: {abs(y_test[i][0] - y_hat_test[i][0]):.2f}")
```

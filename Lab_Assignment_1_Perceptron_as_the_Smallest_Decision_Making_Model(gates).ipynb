```python
import numpy as np

class Perceptron:
    def __init__(self, input_size, lr=0.1, epochs=10):
        self.w = np.zeros(input_size)
        self.b = 0.0
        self.lr = lr
        self.epochs = epochs

    def activation(self, z):
        return 1 if z >= 0 else 0

    def predict(self, x):
        z = np.dot(x, self.w) + self.b
        return self.activation(z)

    def train(self, X, y):
        for epoch in range(self.epochs):
            for i in range(len(X)):
                y_hat = self.predict(X[i])
                error = y[i] - y_hat

                if error != 0:
                    self.w += self.lr * error * X[i]
                    self.b += self.lr * error



gates = {
    "AND":  (np.array([[0,0], [0,1], [1,0], [1,1]]), np.array([0, 0, 0, 1])),
    "OR":   (np.array([[0,0], [0,1], [1,0], [1,1]]), np.array([0, 1, 1, 1])),
    "NAND": (np.array([[0,0], [0,1], [1,0], [1,1]]), np.array([1, 1, 1, 0])),
    "NOR":  (np.array([[0,0], [0,1], [1,0], [1,1]]), np.array([1, 0, 0, 0])),
    "XOR":  (np.array([[0,0], [0,1], [1,0], [1,1]]), np.array([0, 1, 1, 0])) # For observation [cite: 485]
}

for gate_name, (X, y) in gates.items():
    model = Perceptron(input_size=2, lr=0.1, epochs=20)
    model.train(X, y)
    print(f"--- {gate_name} Gate ---")
    for i in range(len(X)):
        print(f"Input: {X[i]}, Predicted: {model.predict(X[i])}")
    print(f"Final Weights: {model.w}, Final Bias: {model.b}\n") # [cite: 487, 488]
```

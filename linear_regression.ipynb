```python
import numpy as np
import matplotlib.pyplot as plt

# Features (X) and Target (y)
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([5, 7, 9, 11, 13], dtype=float)
```

```python
def compute_cost(X, y, w, b):
    m = len(X)
    predictions = w * X + b
    cost = (1 / (2 * m)) * np.sum((predictions - y)**2)
    return cost
```

```python
def compute_gradient(X, y, w, b):
    m = len(X)
    dj_dw = (1 / m) * np.sum(((w * X + b) - y) * X)
    dj_db = (1 / m) * np.sum((w * X + b) - y)
    return dj_dw, dj_db
```

```python
w, b = 0.0, 0.0
alpha = 0.01
epochs = 2000

for i in range(epochs):
    dw, db = compute_gradient(X, y, w, b)

    # Update parameters
    w = w - alpha * dw
    b = b - alpha * db

    # Print progress every 200 steps
    if i % 200 == 0:
        print(f"Epoch {i}: Cost = {compute_cost(X, y, w, b):.4f}, w = {w:.2f}, b = {b:.2f}")

print(f"\nFinal Model: f(x) = {w:.2f}x + {b:.2f}")
```

```python
def predict(x_val):
    return w * x_val + b

# Plotting the results
plt.scatter(X, y, color='red', label='Original Data')
plt.plot(X, w*X + b, color='blue', label='ML Regression Line')
plt.legend()
plt.show()

print(f"Prediction for x=10: {predict(10)}")
```
